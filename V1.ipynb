```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, PolynomialFeatures
from sklearn.cluster import KMeans



# 1. Cyclical Transformation
X['Age_sin'] = -np.sin(2 * np.pi * X['Age'] / 100)  # Assume age periodicity is 100 years
X['Age_cos'] = np.cos(2 * np.pi * X['Age'] / 100)

# 2. Age Relative to Specific Cutoffs
X['Age_Above_18'] = X['Age'] - 18
X['Age_Under_60'] = -(60 - X['Age'])
X['Age_Midpoint'] = np.abs(X['Age'] - 40)  # Distance from a meaningful midpoint

# 3. Age Clustering
kmeans = KMeans(n_clusters=3, random_state=42)
X['Age_Cluster'] = kmeans.fit_predict(X[['Age']])


# 4. Log Transformation
X['Age_Log'] = np.log1p(X['Age'])  # Use log1p to avoid log(0)

# 5. Scaled Binning (0 to 1)
bins = [0, 20, 40, 60, 80]
X['Age_Binned_Scaled'] = pd.cut(X['Age'], bins=bins, labels=[0, 0.33, 0.67, 1])


# 7. Polynomial Features (scaled 0 to 1)
poly = PolynomialFeatures(degree=2, include_bias=False)
age_poly = poly.fit_transform(X[['Age']])
# Scale polynomial features
age_poly_scaled = MinMaxScaler().fit_transform(age_poly)
poly_features = pd.DataFrame(age_poly_scaled, columns=['Age_Scaled', 'Age^2'])
X = pd.concat([X, poly_features], axis=1)



```

```python
# Function to apply transformations to a single feature
def transform_feature(X, feature_name):
    scaler = MinMaxScaler()
    poly = PolynomialFeatures(degree=2, include_bias=False)

    # 1. Scale the feature to [0, 1]
    X[f'{feature_name}_Scaled'] = scaler.fit_transform(X[[feature_name]])

    # 2. Polynomial feature (squared term)
    feature_poly = poly.fit_transform(X[[f'{feature_name}_Scaled']])
    X[f'{feature_name}^2'] = feature_poly[:, 1]  # Extract squared term

    # 3. Sine transformation
    X[f'{feature_name}_Sin'] = -np.sin(2 * np.pi * X[f'{feature_name}_Scaled'])

    # 4. Cosine transformation
    X[f'{feature_name}_Cos'] = -np.cos(2 * np.pi * X[f'{feature_name}_Scaled'])

    # 5. Log transformation (adding epsilon to avoid log(0))
    epsilon = 1e-10
    X[f'{feature_name}_Log'] = np.log(X[feature_name] + epsilon)
    X[f'{feature_name}_Log_Scaled'] = scaler.fit_transform(X[[f'{feature_name}_Log']])

    # 6. Clustering
    kmeans = KMeans(n_clusters=3, random_state=42)
    X[f'{feature_name}_Cluster'] = kmeans.fit_predict(X[[feature_name]])

    return X

# Apply transformations to each feature
features_to_transform = [
    'Speed_Difference',
    'Speed_Ratio',
    'Road_Risk_Index',
    'LCE',
    'Total_Lane_width',
    'Lane_Utilization'
]

for feature in features_to_transform:
    X = transform_feature(X, feature)
```

```python
# Required Libraries
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
from keras.models import Sequential
from keras.layers import Dense, Dropout
from keras.optimizers import Adam, Nadam, RMSprop
from keras.regularizers import l2
from scikeras.wrappers import KerasClassifier
from keras.callbacks import EarlyStopping
from sklearn.model_selection import RandomizedSearchCV
import keras_tuner as kt

# Encode target labels if categorical
encoder = LabelEncoder()
y_train_encoded = encoder.fit_transform(y_train_balanced)
y_test_encoded = encoder.transform(y_test_balanced)

# Standardize features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train_balanced)
X_test = scaler.transform(X_test_balanced)

def build_model(hp):
    model = Sequential()
    model.add(Dense(
        units=hp.Int('units_layer1', min_value=64, max_value=256, step=64),
        activation='relu',
        input_dim=X_train.shape[1],
        kernel_regularizer=l2(hp.Choice('l2_regularizer', values=[0.01, 0.001, 0.0001]))
    ))
    model.add(Dropout(hp.Float('dropout_rate_layer1', min_value=0.2, max_value=0.5, step=0.05)))

    model.add(Dense(
        units=hp.Int('units_layer2', min_value=64, max_value=256, step=64),
        activation='relu',
        kernel_regularizer=l2(hp.Choice('l2_regularizer', values=[0.01, 0.001, 0.0001]))
    ))
    model.add(Dropout(hp.Float('dropout_rate_layer2', min_value=0.2, max_value=0.5, step=0.05)))

    model.add(Dense(len(np.unique(y_train_encoded)), activation='softmax'))

    model.compile(
        optimizer=hp.Choice('optimizer', values=['adam', 'nadam', 'rmsprop']),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

# Initialize the tuner
tuner = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=30,  # Reduce trials to focus on quality
    executions_per_trial=3,
    directory='my_dir',
    project_name='optimized_keras_tuner'
)

# Early stopping to prevent overfitting
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)

# Search for the best hyperparameters
tuner.search(
    X_train, y_train_encoded,
    epochs=200,  # Reduce epochs to avoid overfitting
    validation_data=(X_test, y_test_encoded),
    callbacks=[early_stopping]
)

# Get the best model
best_model = tuner.get_best_models(num_models=1)[0]

# Evaluate the model
y_pred = np.argmax(best_model.predict(X_test), axis=1)

print("Classification Report:")
print(classification_report(y_test_encoded, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test_encoded, y_pred))


# Get feature importance based on the first dense layer's weights
first_layer_weights = best_model.layers[0].get_weights()[0]  # Weights of the first Dense layer
feature_importance = np.mean(np.abs(first_layer_weights), axis=1)  # Mean of absolute weights

# Create a DataFrame for feature importance
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': feature_importance
}).sort_values(by='Importance', ascending=False)

# Plot all features
plt.figure(figsize=(20, 15))
plt.barh(importance_df['Feature'], importance_df['Importance'], color='skyblue')
plt.xlabel('Feature Importance')
plt.ylabel('Feature Names')
plt.title('Feature Importance for Classification')
plt.gca().invert_yaxis()  # Invert y-axis for better readability
plt.show()

```
